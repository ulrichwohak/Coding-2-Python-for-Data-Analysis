# Session 14 Exercise: Categorical Variables And Interactions

Fit a model with sector effects, then add an interaction between training hours and sector.


```python
import pandas as pd
import statsmodels.formula.api as smf

from pathlib import Path


def find_repo_root(start=Path.cwd()):
    for path in [start, *start.parents]:
        if (path / "pyproject.toml").exists():
            return path
    raise RuntimeError("Could not find repository root")


ROOT = find_repo_root()
DATA = ROOT / "data" / "raw"
```

```python
df = pd.read_csv(DATA / "career_outcomes.csv")
without_interaction = smf.ols(
    "salary_k_eur ~ experience_years + training_hours + C(sector)",
    data=df,
).fit()
with_interaction = smf.ols(
    "salary_k_eur ~ experience_years + training_hours + C(sector) + training_hours:C(sector)",
    data=df,
).fit()
pd.DataFrame(
    {
        "model": ["without_interaction", "with_interaction"],
        "adjusted_r_squared": [without_interaction.rsquared_adj, with_interaction.rsquared_adj],
    }
)
```
