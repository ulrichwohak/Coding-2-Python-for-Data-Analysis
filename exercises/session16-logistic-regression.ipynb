# Session 16 Exercise: Logistic Regression

Fit a logistic model for campaign response and compare two classification thresholds.


```python
import pandas as pd
import statsmodels.formula.api as smf
from sklearn.metrics import confusion_matrix

from pathlib import Path


def find_repo_root(start=Path.cwd()):
    for path in [start, *start.parents]:
        if (path / "pyproject.toml").exists():
            return path
    raise RuntimeError("Could not find repository root")


ROOT = find_repo_root()
DATA = ROOT / "data" / "raw"
```

```python
campaign = pd.read_csv(DATA / "campaign_response.csv")
model = smf.logit(
    "responded ~ visits_last_month + emails_opened + discount_pct + prior_spend_eur + C(segment)",
    data=campaign,
).fit(disp=False)
campaign = campaign.assign(probability=model.predict(campaign))
```

```python
rows = []
for threshold in [0.4, 0.5, 0.6]:
    predicted = (campaign["probability"] >= threshold).astype(int)
    tn, fp, fn, tp = confusion_matrix(campaign["responded"], predicted).ravel()
    rows.append({"threshold": threshold, "true_positive": tp, "false_positive": fp, "false_negative": fn})
pd.DataFrame(rows)
```

Which threshold would you choose if false positives are expensive? Which would you choose if missed responders are expensive?
