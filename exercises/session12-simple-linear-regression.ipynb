# Session 12 Exercise: Simple Linear Regression

Fit a simple regression that predicts `salary_k_eur` from one numeric predictor. Then write a two-sentence interpretation of the slope.


```python
import pandas as pd
import statsmodels.formula.api as smf

from pathlib import Path


def find_repo_root(start=Path.cwd()):
    for path in [start, *start.parents]:
        if (path / "pyproject.toml").exists():
            return path
    raise RuntimeError("Could not find repository root")


ROOT = find_repo_root()
DATA = ROOT / "data" / "raw"
```

```python
df = pd.read_csv(DATA / "career_outcomes.csv")
predictor = "training_hours"
model = smf.ols(f"salary_k_eur ~ {predictor}", data=df).fit()
print(model.params)
```

Try a different numeric predictor. Which model has a clearer substantive interpretation? Which has the higher R-squared?


```python
candidate_predictors = ["experience_years", "education_years", "ai_tool_use"]
summary_rows = []
for name in candidate_predictors:
    fitted = smf.ols(f"salary_k_eur ~ {name}", data=df).fit()
    summary_rows.append({"predictor": name, "slope": fitted.params[name], "r_squared": fitted.rsquared})
pd.DataFrame(summary_rows)
```
