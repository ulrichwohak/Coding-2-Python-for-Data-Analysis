# Session 13 Exercise: Multiple Regression Workflow

Compare two predictor sets for salary. Use the same train/test split for both models.


```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
from sklearn.model_selection import train_test_split

from pathlib import Path


def find_repo_root(start=Path.cwd()):
    for path in [start, *start.parents]:
        if (path / "pyproject.toml").exists():
            return path
    raise RuntimeError("Could not find repository root")


ROOT = find_repo_root()
DATA = ROOT / "data" / "raw"
```

```python
df = pd.read_csv(DATA / "career_outcomes.csv")
y = df["salary_k_eur"]
feature_sets = {
    "baseline": ["training_hours"],
    "expanded": ["training_hours", "experience_years", "education_years", "ai_tool_use"],
}
```

```python
rows = []
for label, features in feature_sets.items():
    X_train, X_test, y_train, y_test = train_test_split(df[features], y, test_size=0.25, random_state=42)
    model = LinearRegression().fit(X_train, y_train)
    pred = model.predict(X_test)
    rows.append({"model": label, "test_rmse": mean_squared_error(y_test, pred) ** 0.5})
pd.DataFrame(rows)
```
