# Session 15 Exercise: Diagnostics And Transformations

Fit a housing price model, inspect residuals, and compare a log-price model.


```python
import numpy as np
import pandas as pd
import statsmodels.formula.api as smf

from pathlib import Path


def find_repo_root(start=Path.cwd()):
    for path in [start, *start.parents]:
        if (path / "pyproject.toml").exists():
            return path
    raise RuntimeError("Could not find repository root")


ROOT = find_repo_root()
DATA = ROOT / "data" / "raw"
```

```python
housing = pd.read_csv(DATA / "housing_sales.csv")
formula = "price_k_eur ~ size_sq_m + rooms + age_years + renovation_score + C(district)"
linear = smf.ols(formula, data=housing).fit()
log_price = smf.ols(formula.replace("price_k_eur", "np.log(price_k_eur)"), data=housing).fit()
pd.DataFrame(
    {
        "model": ["linear", "log_price"],
        "adjusted_r_squared": [linear.rsquared_adj, log_price.rsquared_adj],
        "aic": [linear.aic, log_price.aic],
    }
)
```

```python
diagnostics = housing.assign(fitted=linear.fittedvalues, residual=linear.resid)
diagnostics[["fitted", "residual"]].head()
```
